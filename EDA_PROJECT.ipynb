```python
Q1 = df['unit_price_usd'].quantile(0.25)
Q3 = df['unit_price_usd'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = ((df['unit_price_usd'] < lower) | 
            (df['unit_price_usd'] > upper)).sum()

print("Number of outliers:", outliers)
```

```python
df[df['unit_price_usd'] > upper]['unit_price_usd'].describe()
```

```python
df[df['unit_price_usd'] > upper][
    ['product_name', 'category', 'unit_price_usd']
].head(10)
```

```python
outlier_cols = [
    'age',
    'customer_loyalty_score',
    'total_orders_by_customer',
    'product_reviews_count',
    'stock_quantity',
    'unit_price_usd',
    'quantity',
    'discount_percent',
    'total_price_usd',
    'cost_usd',
    'profit_usd',
    'tax_usd',
    'shipping_cost_usd',
    'delivery_days',
    'session_duration_minutes',
    'pages_visited',
    'profit_margin_percent'
]

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    
    print(col, ":", count)
```

```python
outlier_profit = df[
    df['profit_margin_percent'] > 50
]

outlier_profit[
    ['unit_price_usd',
     'cost_usd',
     'profit_usd',
     'profit_margin_percent']
].head(10)
```

```python
df['calculated_total'] = (
    df['unit_price_usd'] * df['quantity']
) - df['discount_amount_usd']

df['price_difference'] = (
    df['total_price_usd'] - df['calculated_total']
)

df['price_difference'].describe()
```

```python
(df['price_difference'].abs() > 0.01).sum()
```

```python
df['calculated_profit'] = (
    df['total_price_usd'] - df['cost_usd']
)

df['profit_difference'] = (
    df['profit_usd'] - df['calculated_profit']
)

(df['profit_difference'].abs() > 0.01).sum()
```

```python
df.drop(
    columns=['calculated_total',
             'price_difference',
             'calculated_profit',
             'profit_difference'],
    inplace=True
)
```
